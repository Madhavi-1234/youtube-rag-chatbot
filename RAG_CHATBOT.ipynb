```python
db = None
text = None
chunks = None

def process_video(video_url):
    global db, text, chunks

    # already processed
    if db is not None:
        return db

    # 1. download audio
    import yt_dlp

    ydl_opts = {
        'format': 'bestaudio/best',
        'outtmpl': 'audio',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
        }]
    }

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([video_url])

    # 2. transcription (FASTER WHISPER)
    from faster_whisper import WhisperModel

    model = WhisperModel("tiny.en", device="cpu")

    segments, info = model.transcribe("audio.mp3")

    text = ""

    for segment in segments:
        text += segment.text + " "

    # 3. chunking
    from langchain_text_splitters import RecursiveCharacterTextSplitter

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=300,
        chunk_overlap=30
    )

    chunks = splitter.create_documents([text])

    # 4. embeddings + FAISS
    from langchain_community.embeddings import HuggingFaceEmbeddings
    from langchain_community.vectorstores import FAISS

    emb = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    db = FAISS.from_documents(chunks, emb)

    return db
```

```python
import gradio as gr

# ensure DB global
db = None

def run_pipeline(video_url, query):

    global db

    # process video only once
    if db is None:
        db = process_video(video_url)

    # retrieve docs
    docs = db.similarity_search(query, k=3)
    context = "\n".join([d.page_content for d in docs])

    # Gemini response
    import google.generativeai as genai

    model = genai.GenerativeModel("gemini-2.5-flash")

    prompt = f"""
Answer ONLY from the context.

Context:
{context}

Question:
{query}

If not found say: Not available in video.
"""

    response = model.generate_content(prompt)

    return response.text
```

```python
interface = gr.Interface(
    fn=run_pipeline,
    inputs=[
        gr.Textbox(label="YouTube Video URL"),
        gr.Textbox(label="Ask Question")
    ],
    outputs="text",
    title="🎥 YouTube RAG Chatbot (Whisper + FAISS + Gemini)"
)

interface.launch(share=True)
```
